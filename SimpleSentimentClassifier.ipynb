```python
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.naive_bayes import MultinomialNB
# 1. Training data: sentences + their labels
texts = [
    "I love this product",
    "This is amazing",
    "Best purchase ever",
    "I am very happy with this",
    "Great quality and fast delivery",
    "I hate this product",
    "This is terrible",
    "Worst purchase ever",
    "I am very disappointed",
    "Poor quality and slow delivery",
]

labels = [
    "Positive", "Positive", "Positive", "Positive", "Positive",
    "Negative", "Negative", "Negative", "Negative", "Negative",
]

# 2. Convert text into numeric features (word counts)
vectorizer = CountVectorizer()
X = vectorizer.fit_transform(texts)

# 3. Train the model
model = MultinomialNB()
model.fit(X, labels)

# 4. Test it on new sentences
new_sentences = [
    "I really love this",
    "This is the worst thing I bought",
    "Fast delivery and great quality",
]

new_X = vectorizer.transform(new_sentences)
predictions = model.predict(new_X)

for sentence, prediction in zip(new_sentences, predictions):
    print(f"'{sentence}' -> {prediction}")
```
